## Dataset Documentation
This lab utilizes the llm_hallucination_dataset_v1.csv dataset, which contains 200 text-based records capturing prompts provided to various Large Language Models (LLMs) and their subsequent generated responses. Key columns include unstructured text data such as prompt_text, response_text, and correction_text, alongside categorical data indicating the domain, task_type, and a binary hallucination_label (1 for a hallucination, 0 otherwise). This dataset is well-documented and highly appropriate for exploring text length distributions, token frequencies, and domain representations in natural language processing tasks.
## Python Implementation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import re

# Set plotting style for clean visuals
sns.set_theme(style="whitegrid")

# Load the text dataset
df = pd.read_csv('llm_hallucination_dataset_v1.csv')

# Compute text characteristics: Word counts for prompts and responses
df['prompt_word_count'] = df['prompt_text'].apply(lambda x: len(str(x).split()))
df['response_word_count'] = df['response_text'].apply(lambda x: len(str(x).split()))

# Visualization 1
# Create a map for clear labeling
df['hallucination_status'] = df['hallucination_label'].map({1: 'Hallucination', 0: 'No Hallucination'})

# Generate the plot
fig1, ax1 = plt.subplots(figsize=(8, 5))
sns.boxplot(data=df, x='hallucination_status', y='response_word_count', palette='Set2', ax=ax1)

ax1.set_title('LLM Response Word Count Distribution by Hallucination Status', fontsize=14, pad=15)
ax1.set_xlabel('Hallucination Label', fontsize=12)
ax1.set_ylabel('Response Word Count (Number of Words)', fontsize=12)

plt.tight_layout()
fig1.savefig('response_length_by_hallucination.png', dpi=300)

# Visualization 2
# Clean and tokenize prompt text to find frequent descriptive words
words = []
stop_words = set(['the', 'a', 'an', 'and', 'or', 'but', 'is', 'are', 'was', 'were', 'to', 'of', 'in', 'for', 'on', 'with', 
                  'what', 'who', 'how', 'when', 'where', 'why', 'which', 'it', 'by', 'that', 'this', 'from', 'at', 'be', 
                  'can', 'not', 'as', 'your', 'you', 'has', 'have', 'about', 'more', 'many', 'did', 'does', 'type', 'had'])

for text in df['prompt_text']:
    cleaned = re.sub(r'[^\w\s]', '', str(text).lower())
    for word in cleaned.split():
        if word not in stop_words and len(word) > 2:
            words.append(word)

# Aggregate and sort counts
word_counts = Counter(words).most_common(15)
word_df = pd.DataFrame(word_counts, columns=['Word', 'Frequency']).sort_values(by='Frequency', ascending=True)

# Generate the sorted horizontal bar chart
fig2, ax2 = plt.subplots(figsize=(8, 5))
ax2.barh(word_df['Word'], word_df['Frequency'], color='skyblue', edgecolor='gray')

ax2.set_title('Top 15 Most Frequent Content Words in Prompt Text', fontsize=14, pad=15)
ax2.set_xlabel('Frequency (Count)', fontsize=12)
ax2.set_ylabel('Words', fontsize=12)

plt.tight_layout()
fig2.savefig('top_prompt_words.png', dpi=300)

# Visualization 3
# Aggregate domain frequencies and sort in ascending order for clean presentation
domain_counts = df['domain'].value_counts().reset_index()
domain_counts.columns = ['Domain', 'Count']
domain_counts = domain_counts.sort_values(by='Count', ascending=True)

# Generate the sorted horizontal bar chart
fig3, ax3 = plt.subplots(figsize=(8, 5))
ax3.barh(domain_counts['Domain'], domain_counts['Count'], color='salmon', edgecolor='gray')

ax3.set_title('Frequency Distribution of Text Across Knowledge Domains', fontsize=14, pad=15)
ax3.set_xlabel('Number of Records', fontsize=12)
ax3.set_ylabel('Domain', fontsize=12)

plt.tight_layout()
fig3.savefig('domain_distribution.png', dpi=300)
```

### Visualization 1 Interpretation
* **Figure Explanation:** This boxplot displays the text length distribution (measured in word count) of the generated LLM responses, separated by whether the response contained a hallucination or remained factually accurate.
* **Insight & Interpretation:** The plot illustrates that longer generated response texts are more prone to factual errors, with hallucinated texts exhibiting a significantly higher average length (approximately 35.4 words) compared to accurate responses (approximately 26.3 words).

# Lab: Assessing the Impact of Artificial Intelligence on the Future Labor Market (2030)
### Summary
The dataset I've obtained from Kaggle, AI_Impact_on_Jobs_2030.csv, represents a comprehensive, workforce-wide predictive simulation mapping how artificial intelligence, evolving corporate technology adoption, and automation are anticipated to structurally reshape the global employment market leading up to the year 2030. The dataset contains a total of 3,000 unique professional profile entries. Each profile entry tracks 20 distinct feature columns: Employee_ID, Job_Title, Industry, Country, Education_Level, Years_Experience, AI_Replacement_Risk, Future_Demand_Score, Remote_Work_Possibility, Average_Salary_USD, Required_Skills, Automation_Level, Job_Growth_2030, Work_Hours_Per_Week, Company_Size, AI_Tool_Usage, Performance_Score, Upskilling_Needed, Job_Satisfaction, and Hiring_Trend_2026. Structurally, the columns are cleanly split among three data types: continuous numeric floats representing scores or indices (AI_Replacement_Risk, Future_Demand_Score, Performance_Score, Job_Satisfaction), discrete integers representing quantities, wages, or growth percentages (Years_Experience, Average_Salary_USD, Job_Growth_2030, Work_Hours_Per_Week), and qualitative object strings representing categories or text arrays (Employee_ID, Job_Title, Industry, Country, Education_Level, Remote_Work_Possibility, Required_Skills, Automation_Level, Company_Size, AI_Tool_Usage, Upskilling_Needed, Hiring_Trend_2026). Finally, a systematic verification of the data matrix reveals that there is absolutely no missing data present anywhere in the dataset; every feature column is completely populated across all 3,000 entries, rendering data imputation or null value treatment unnecessary prior to analytical evaluation.

### 5 interesting questions using Pandas methods
**Question 1:**
Which 5 specific industries face the highest average AI Replacement Risk, and what is that risk value?

**Question 2:**
Is there a relationship between a professional's education level and their future demand score? What is the mean Future Demand Score grouped by Education Level?

**Question 3:**
What are the top 5 highest-paying job titles that simultaneously maintain a "Low" Automation Level?

**Question 4:**
How does the hiring outlook change based on employee adaptation? What is the distribution count of 'Hiring_Trend_2026' for employees where 'Upskilling_Needed' equals 'Yes'?

**Question 5:**
For positions characterized by long hours (working greater than 45 hours per week), what are the average job satisfaction levels across different company sizes?


```python
import pandas as pd

# Load the dataset
df = pd.read_csv('AI_Impact_on_Jobs_2030.csv')

print("="*60)
print("      EXECUTING LAB ASSIGNMENT DATA ANALYSIS QUERIES        ")
print("="*60)

# ----------------------------------------------------
# 1: Top 5 Industries by Mean AI Replacement Risk
# ----------------------------------------------------
print("\n[ANSWER 1] Top 5 Industries Facing Highest AI Replacement Risk:")
top_risk_industries = df.groupby('Industry')['AI_Replacement_Risk'].mean().sort_values(ascending=False).head(5)
print(top_risk_industries.to_string())

"""
[ANSWER 1] Top 5 Industries Facing Highest AI Replacement Risk:
Industry
Finance       0.530484
Government    0.528793
Technology    0.517423
Energy        0.499164
Education     0.496933
"""

# ----------------------------------------------------
# 2: Mean Future Demand Score by Education Level
# ----------------------------------------------------
print("\n" + "-"*50)
print("[ANSWER 2] Mean Future Demand Score by Education Level:")
demand_by_education = df.groupby('Education_Level')['Future_Demand_Score'].mean().sort_values(ascending=False)
print(demand_by_education.to_string())

"""
[ANSWER 2] Mean Future Demand Score by Education Level:
Education_Level
High School    0.611596
PhD            0.603582
Bachelor       0.603095
Master         0.585913
"""

# ----------------------------------------------------
# 3: Top 5 Highest-Paying Jobs with Low Automation Risk
# ----------------------------------------------------
print("\n" + "-"*50)
print("[ANSWER 3] Top 5 Highest-Paying Job Titles with 'Low' Automation Level:")
high_pay_low_risk = df[df['Automation_Level'] == 'Low'].groupby('Job_Title')['Average_Salary_USD'].mean().sort_values(ascending=False).head(5)
print(high_pay_low_risk.to_string())

"""
[ANSWER 3] Top 5 Highest-Paying Job Titles with 'Low' Automation Level:
Job_Title
IoT Specialist      142357.186441
HR Specialist       140016.980769
Digital Marketer    135669.705882
Prompt Engineer     131735.254545
Business Analyst    130964.836735
"""

# ----------------------------------------------------
# 4: Hiring Trend Distribution for Workers Needing Upskilling
# ----------------------------------------------------
print("\n" + "-"*50)
print("[ANSWER 4] 2026 Hiring Trends for Employees Who Need Upskilling:")
upskilling_trends = df[df['Upskilling_Needed'] == 'Yes']['Hiring_Trend_2026'].value_counts()
print(upskilling_trends.to_string())

"""
[ANSWER 4] 2026 Hiring Trends for Employees Who Need Upskilling:
Hiring_Trend_2026
Declining    502
Growing      497
Stable       490
"""

# ----------------------------------------------------
# 5: Average Job Satisfaction for Overworked Personnel (>45 hrs/wk)
# ----------------------------------------------------
print("\n" + "-"*50)
print("[ANSWER 5] Avg Job Satisfaction for Hard Workers (>45 Hours/Week) by Company Size:")
satisfaction_overworked = df[df['Work_Hours_Per_Week'] > 45].groupby('Company_Size')['Job_Satisfaction'].mean().sort_values(ascending=False)
print(satisfaction_overworked.to_string())
print("="*60)

"""
[ANSWER 5] Avg Job Satisfaction for Hard Workers (>45 Hours/Week) by Company Size:
Company_Size
Medium        3.041653
Enterprise    3.041305
Startup       3.038547
"""
print("")
```

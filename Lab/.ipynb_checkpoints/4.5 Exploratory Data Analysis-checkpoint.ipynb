# 4.5 EDA and Preparing Data for Machine Learning Models
## Dataset Overview & Feature Documentation
**Dataset Description**

The objective of this exploratory data analysis (EDA) is to understand the factors driving consumer spending behavior on an e-commerce platform. By selecting an appropriate target feature and analyzing its relationships with various consumer attributes, we aim to uncover key insights that can prepare this data for future predictive machine learning models.

**Target Feature**
* **Purchase_Amount (Numerical):** The total dollar amount spent by the customer on a given transaction. This serves as our primary continuous metric for business value.

**Input Features**
* **Age (Numerical):** The age of the customer in years. Used to identify generational spending behavior.
* **Income_Level (Categorical):** Segmented into 'Low', 'Middle', and 'High'. Used to determine if economic bracket dictates transaction size.
* **Frequency_of_Purchase (Numerical):** The number of times the customer purchases from the platform in a given timeframe. Identifies if highly frequent shoppers buy more expensive items or micro-items.
* **Discount_Sensitivity (Categorical):** Categorized into 'Not Sensitive', 'Somewhat Sensitive', and 'Highly Sensitive'. Measures how much bargain-hunting influences total checkout amounts.
* **Customer_Loyalty_Program_Member (Boolean):** Indicates whether the user is enrolled in the platform's loyalty program (True/False). Evaluates the financial efficacy of the membership model.

## Python Implmentation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting aesthetics
sns.set_theme(style="whitegrid")
plt.rcParams.update({
    'font.size': 12,
    'axes.labelsize': 14,
    'axes.titlesize': 16,
    'xtick.labelsize': 12,
    'ytick.labelsize': 12,
    'figure.titlesize': 18
})

# Load Dataset
df = pd.read_csv('Ecommerce_Consumer_Behavior_Analysis_Data.csv')

# Data Cleaning: Convert Target Feature 'Purchase_Amount' to Numeric
# Strip out '$' sign and convert to float
df['Purchase_Amount'] = df['Purchase_Amount'].astype(str).str.replace('$', '', regex=False).astype(float)

# Filter down to the features required
selected_features = [
    'Purchase_Amount', 
    'Age', 
    'Income_Level', 
    'Frequency_of_Purchase', 
    'Discount_Sensitivity', 
    'Customer_Loyalty_Program_Member'
]
df_clean = df[selected_features].copy()
df_clean['Income_Level'] = df_clean['Income_Level'].fillna('Unknown')
df_clean['Discount_Sensitivity'] = df_clean['Discount_Sensitivity'].fillna('Unknown')

# Visualization 1
plt.figure(figsize=(9, 5))
sns.regplot(
    data=df_clean, 
    x='Age', 
    y='Purchase_Amount', 
    scatter_kws={'alpha':0.4, 'color': '#2b5c8f'}, 
    line_kws={'color': '#d95f02', 'linewidth': 2}
)
plt.title('Relationship Between Customer Age and Purchase Amount', pad=15)
plt.xlabel('Customer Age (Years)')
plt.ylabel('Purchase Amount ($)')
plt.tight_layout()
plt.show()

# Visualization 2
plt.figure(figsize=(9, 5))
# Force logical order of categories
income_order = ['Low', 'Middle', 'High']

sns.boxplot(
    data=df_clean, 
    x='Income_Level', 
    y='Purchase_Amount', 
    order=income_order,
    palette='Blues',
    width=0.5
)
plt.title('Purchase Amount Distribution Across Income Levels', pad=15)
plt.xlabel('Income Level')
plt.ylabel('Purchase Amount ($)')
plt.tight_layout()
plt.show()

# Visualization 3
plt.figure(figsize=(9, 5))
sns.lineplot(
    data=df_clean, 
    x='Frequency_of_Purchase', 
    y='Purchase_Amount', 
    color='#2ca02c', 
    linewidth=2.5,
    errorbar=None, # Focus on trends
    marker='o'
)
plt.title('Average Purchase Amount by Purchase Frequency', pad=15)
plt.xlabel('Frequency of Purchase (Count)')
plt.ylabel('Average Purchase Amount ($)')
plt.xticks(sorted(df_clean['Frequency_of_Purchase'].unique()))
plt.tight_layout()
plt.show()

# Visualization 4
plt.figure(figsize=(9, 5))
sensitivity_order = ['Not Sensitive', 'Somewhat Sensitive', 'Highly Sensitive']

sns.violinplot(
    data=df_clean, 
    x='Discount_Sensitivity', 
    y='Purchase_Amount', 
    order=sensitivity_order,
    palette='muted',
    inner='quartile'
)
plt.title('Purchase Amount Density by Discount Sensitivity', pad=15)
plt.xlabel('Discount Sensitivity Tier')
plt.ylabel('Purchase Amount ($)')
plt.tight_layout()
plt.show()

# Visualization 5
plt.figure(figsize=(9, 5))
sns.barplot(
    data=df_clean, 
    x='Customer_Loyalty_Program_Member', 
    y='Purchase_Amount', 
    palette='Set2',
    capsize=0.1,
    errorbar='ci'
)
plt.title('Mean Purchase Amount by Loyalty Program Status', pad=15)
plt.xlabel('Customer Loyalty Program Member')
plt.ylabel('Mean Purchase Amount ($)')
plt.xticks([0, 1], ['Non-Member (False)', 'Member (True)'])
plt.tight_layout()
plt.show()
```

## Visualizations & Interpretations

![](a1.png)
### Interpretation 1: Age vs. Purchase Amount
The regression scatter plot illustrates the relationship between consumer age and their checkout purchase amount. The flat slope of the fitted regression line indicates a negligible or non-existent linear relationship between these two variables. Spending is widely distributed between $0 and $500 across all generations, meaning age alone does not dictate financial value per order. This suggests that marketing campaigns targeting specific age cohorts might not see variances in average basket size.
![](a2.png)
### Interpretation 2: Income Level vs. Purchase Amount
This box plot compares order value across different income groups. Interestingly, the median spending (represented by the horizontal line inside the boxes), the interquartile range (IQR), and the minimum/maximum distribution limits remain remarkably consistent across Low, Middle, and High-income consumers. This indicates that income levels do not strongly predict transaction values on this platform. The platform remains equally accessible and fully integrated into the shopping budgets of all economic tiers.
![](a3.png)
### Interpretation 3: Purchase Frequency vs. Purchase Amount
This line plot explores whether high-frequency buyers purchase higher or lower value items on average. The visualization reveals an erratic, fluctuating line chart with no continuous upward or downward trajectory. Shoppers with a frequency of 14 show an average ticket drop, while those at 11 spike. Because no consistent trend emerges, we can determine that a customer's ordering frequency does not reliably scale or diminish the baseline dollar amount they spend per individual order.
![](a4.png)
### Interpretation 4: Discount Sensitivity vs. Purchase Amount
The violin plot visualizes the shape and density distribution of spending across discount-sensitivity classes. The overall distribution profiles are uniform: wide, balanced shapes across all categories indicate that order values are relatively distributed evenly from low to high thresholds. Whether a consumer actively stalks discounts ('Highly Sensitive') or completely ignores bargains ('Not Sensitive'), the final invoice amounts display virtually identical distributions. Bargain-hunting behavior impacts promotional usage, but it does not diminish total revenue per transaction.
![](a5.png)
### Interpretation 5: Loyalty Program Membership vs. Purchase Amount
This bar chart visualizes the average spending differences between platform loyalty members and non-members, including error bars to signify confidence intervals. The chart reveals that both bars sit at roughly identical heights (~$250 per transaction), and the overlapping error confidence boundaries suggest that there is no statistically significant difference in order size between loyalty club members and general consumers.

## Summary & Conclusion

### Which features seem most related?
* None. Surprisingly, none of the 5 selected features (Age, Income_Level, Frequency_of_Purchase, Discount_Sensitivity, or Customer_Loyalty_Program_Member) show a strong statistical correlation or variance influence on the total Purchase_Amount. 

### Which features don't seem to influence the target?
* All 5 evaluated demographic and behavioral categories yield a uniform distribution across transaction amounts. The average basket size remains centered near $250, regardless of whether a customer is young or old, low or high income, a deal finder, or an enrolled loyalty member.

### The Big Picture Takeaway
From a machine learning preparation standpoint, this indicates that the variation in order values is not driven by simple baseline demographics or membership behaviors. Linear models utilizing these features exclusively will struggle with underfitting. 
To build a high-performing predictive model for Purchase_Amount, future engineering must look to features outside this subset such as specific Purchase_Category, Product_Rating, or structural behavioral elements like Time_Spent_on_Product_Research(hours) to unlock predictive capabilities.

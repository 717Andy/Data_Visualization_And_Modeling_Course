# 2.3 Visualizing Relationships Between Categorical Features Lab

## Python implementation 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('titanic_passengers.csv')

# Categorize Age using .loc
df['Age_Group'] = 'Unknown'
df.loc[df['Age'] < 12, 'Age_Group'] = 'Child'
df.loc[(df['Age'] >= 12) & (df['Age'] < 18), 'Age_Group'] = 'Youth'
df.loc[(df['Age'] >= 18) & (df['Age'] < 60), 'Age_Group'] = 'Adult'
df.loc[df['Age'] >= 60, 'Age_Group'] = 'Senior'

# Categorize Fare using .loc based on the df['Fare'].describe() quartiles
# Q1 (25%) = 7.91, Median (50%) = 14.45, Q3 (75%) = 31.00
df['Fare_Category'] = 'Low'
df.loc[(df['Fare'] > 7.91) & (df['Fare'] <= 14.45), 'Fare_Category'] = 'Medium'
df.loc[(df['Fare'] > 14.45) & (df['Fare'] <= 31.00), 'Fare_Category'] = 'High'
df.loc[df['Fare'] > 31.00, 'Fare_Category'] = 'Luxury'

# Combine SibSp and Parch to get total family size for travel companions
df['Family_Size'] = df['SibSp'] + df['Parch']

# Visualizations & Analysis
# Sex vs. Survival
plt.figure(figsize=(6, 4))
sns.barplot(x='Sex', y='Survived', data=df, errorbar=None, palette='Blues_r')
plt.title('Survival Rate by Sex')
plt.ylabel('Proportion Survived')
plt.xlabel('Sex')
plt.ylim(0, 1)
plt.show()

# Passenger Class (Pclass) vs. Survival
plt.figure(figsize=(6, 4))
sns.barplot(x='Pclass', y='Survived', data=df, errorbar=None, palette='magma')
plt.title('Survival Rate by Passenger Class')
plt.ylabel('Proportion Survived')
plt.xlabel('Class (1 = 1st, 2 = 2nd, 3 = 3rd)')
plt.ylim(0, 1)
plt.show()

# Age Group vs. Survival
plt.figure(figsize=(7, 4))
sns.barplot(x='Age_Group', y='Survived', data=df, order=['Child', 'Youth', 'Adult', 'Senior', 'Unknown'], errorbar=None, palette='viridis')
plt.title('Survival Rate by Age Group')
plt.ylabel('Proportion Survived')
plt.xlabel('Age Category')
plt.ylim(0, 1)
plt.show()

# Fare Category vs. Survival
plt.figure(figsize=(7, 4))
sns.barplot(x='Fare_Category', y='Survived', order=['Low', 'Medium', 'High', 'Luxury'], data=df, errorbar=None, palette='crest')
plt.title('Survival Rate by Fare Category')
plt.ylabel('Proportion Survived')
plt.xlabel('Fare Tier')
plt.ylim(0, 1)
plt.show()

# Family Size vs. Survival
plt.figure(figsize=(8, 4))
sns.barplot(x='Family_Size', y='Survived', data=df, errorbar=None, palette='flare')
plt.title('Survival Rate by Family Size')
plt.ylabel('Proportion Survived')
plt.xlabel('Number of Family Members On Board')
plt.ylim(0, 1)
plt.show()

# Port of Departure (Embarked) vs. Survival
plt.figure(figsize=(6, 4))
sns.barplot(x='Embarked', y='Survived', data=df, errorbar=None, palette='Set2')
plt.title('Survival Rate by Port of Departure')
plt.ylabel('Proportion Survived')
plt.xlabel('Port (C = Cherbourg, Q = Queenstown, S = Southampton)')
plt.ylim(0, 1)
plt.show()

# Query for a passenger matching Profile 1 (Most Likely)
survivor_match = df[
    (df['Sex'] == 'female') & 
    (df['Pclass'] == 1) & 
    (df['Embarked'] == 'C') & 
    (df['Survived'] == 1)
]
print("Potential Real Survivors:")
print(survivor_match[['Name', 'Age', 'Fare']].head(3))

print("\n" + "="*50 + "\n")

# Query for a passenger matching Profile 2 (Least Likely)
victim_match = df[
    (df['Sex'] == 'male') & 
    (df['Pclass'] == 3) & 
    (df['Embarked'] == 'S') & 
    (df['Family_Size'] == 0) & 
    (df['Survived'] == 0)
]
print("Potential Real Victims:")
print(victim_match[['Name', 'Age', 'Fare']].head(3))
```

## Categorizing Quantitative Features
### Reasoning for categories:
* **Age:** Separated into developmental stages because children (Age < 12) were given strict priority during evacuation ("women and children first"), while adults and seniors faced different survival rates.
* **Fare:** Bucketed precisely according to the dataset's quartile distribution (25%, 50%, 75%) to objectively divide passengers into economic tiers (Low, Medium, High, Luxury) matching their ticket pricing.
